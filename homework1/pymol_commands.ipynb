```python
select flexible_loop, b > 15
show sticks, flexible_loop
label n. CA and flexible_loop, "%s-%s" % (resn, resi)
zoom flexible_loop
```

# 5. Show the molecular surface

```python
hide everything
show surface
```

# 6. Define Hydrophobic and Polar groups

```python
select hydrophobic_res, resn ALA+GLY+ILE+LEU+MET+PHE+TRP+VAL+PRO
select polar_res, resn ARG+ASN+ASP+CYS+GLU+GLN+HIS+LYS+SER+THR+TYR
```

# 7. Set surface color

```python
set surface_color, default
color orange, hydrophobic_res
color marine, polar_res
```

# 8. Core Residues

```python
select true_core, resi 16+38+47
show sticks, true_core
label n. CA and true_core, "%s-%s" % (resn, resi)
set label_color, white
```

# 9. Surface Residues

```python
hide everything
reinitialize
fetch 2i61, async=0
remove solvent
select surface_set, resn ARG+LYS+ASP+GLU+ASN+GLN+HIS+SER+THR+TYR
show surface
set transparency, 0.4
color white, all
show sticks, surface_set
color marine, surface_set
label n. CA and (resi 28+9+10), "%s-%s" % (resn, resi)
set label_color, white
```

# 10. The i→i+4 Hydrogen Bonding Pattern

```python
hide everything
reinitialize
fetch 2i61, async=0
remove solvent
show sticks, my_helix and name N+CA+C+O
set stick_radius, 0.15
dist hbonds, my_helix and name O, my_helix and name N, 3.2
color yellow, hbonds
hide labels
```

# 11. The Amphipathic Character & Sequence

```python
show sticks, my_helix
hide protons
select h_phobic, my_helix and resn ALA+GLY+VAL+LEU+ILE+PHE+TRP+MET+PRO
color orange, h_phobic
select h_philic, my_helix and resn ARG+LYS+ASP+GLU+ASN+GLN+SER+THR+TYR+HIS+CYS
color marine, h_philic
label n. CA and my_helix, "%s-%s" % (resn, resi)
set label_color, white
```
